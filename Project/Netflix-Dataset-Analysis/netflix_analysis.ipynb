```python
print(df['type'].duplicated().value_counts())
print(df['title'].duplicated().value_counts())
print(df['release_year'].duplicated().value_counts())
df = df.drop_duplicates().drop_duplicates(subset=["title", "type", "release_year"])
df.info()
```

```python
shifted = df["rating"].str.contains("min", na=False) #find the misplaced values
print("\nShifted rows fixed:", shifted.sum())  #sum the misplaced values
df.loc[shifted, "duration"] = df.loc[shifted, "rating"]  #shifing the misplaced values like rating to duration
df.loc[shifted, "rating"] = np.nan # then rating values is become nan
```

```python
df.info()
```

fill the missing values

```python
for i in ['director','cast','country']:
    df[i]=df[i].fillna('Unknown')
```

```python
df['rating']=df['rating'].fillna('Not Rating')
```

```python
df.info()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"].str.strip(),format="mixed")
df["date_added"] = df["date_added"].dt.year.astype("Int64")   # 2021, not 2021.0
# df["date_added"] = df["date_added"].dt.month_name()
```

```python
df = df.dropna(subset=["date_added"]).reset_index(drop=True)
```

```python
# df["date_added"] = df["date_added"].dt.year.astype(int)
```

```python
median_date = df["date_added"].median()
df["date_added"] = df["date_added"].fillna(median_date)
```

```python
df.info()
```

```python
df.loc[:,['country','listed_in']]
```

```python
df["primary_country"] = df["country"].str.split(",").str[0].str.strip() #seperate the country columns like United States, Canada-->United States
df["primary_genre"] = df["listed_in"].str.split(",").str[0].str.strip() #seperate the listed_in columns Dramas, International Movies---->dramas
df["years_to_netflix"] = df["date_added"] - df["release_year"] #relese year and orginal relese year
```

```python
print(df['rating'].unique().sum())
```

TV-Y---->Kids
PG ----->Older Kids
TV-14--->Teens
R ------>Adults
NR ----->Not Rating

```python
rating_map = {
    "TV-Y": "Kids", "TV-Y7": "Kids", "TV-Y7-FV": "Kids", "TV-G": "Kids", "G": "Kids",
    "TV-PG": "Older Kids", "PG": "Older Kids",
    "TV-14": "Teens", "PG-13": "Teens",
    "TV-MA": "Adults", "R": "Adults", "NC-17": "Adults",
    "NR": "Not Rated", "UR": "Not Rated", "Not Rated": "Not Rated",
}
df["audience"] = df["rating"].map(rating_map).fillna("Not Rated")
```

```python
df.info()
```

```python
print("Movies vs TV Shows:", df["type"].value_counts())
print("Top 10 countries:", df[df.primary_country != "Unknown"]["primary_country"].value_counts().head(10))
```

```python
print("Top 10 genres:", df["primary_genre"].value_counts().head(10))
print("Titles added per year:", df["date_added"].value_counts().sort_index())
```

```python
print("Movie duration:", df["duration"].describe())
print("Audience:", df["audience"].value_counts())
```

```python
genres = df["listed_in"].str.split(", ").explode()
print("Top 10 genres (all tags):", genres.value_counts().head(10))
```

```python
actors = df.loc[df.cast != "Unknown", "cast"].str.split(", ").explode()
print("Top 10 actors:", actors.value_counts().head(10))
```

```python
print("\nCleaned shape:", df.shape)
print("Missing values after cleaning:\n", df.isna().sum())
```
